```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("alitaqishah/pakistan-economic-indicators-2000-2025")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# List files in the downloaded directory to find the CSV file
files_in_path = os.listdir(path)
csv_file = None
for f in files_in_path:
    if f.endswith('.csv'):
        csv_file = f
        break

if csv_file:
    full_csv_path = os.path.join(path, csv_file)
    df = pd.read_csv(full_csv_path)
    print(f"Successfully loaded {csv_file} into DataFrame.")
    display(df.head())
else:
    print(f"No CSV file found in {path}")
```

### 1. What was the average GDP growth rate (`gdp_growth_pct`) per decade?

```python
# Group by 'decade' and calculate the mean of 'gdp_growth_pct'
avg_gdp_growth_per_decade = df.groupby('decade')['gdp_growth_pct'].mean().reset_index()
print("Average GDP Growth Rate per Decade:")
display(avg_gdp_growth_per_decade)
```

### 2. Which year had the highest inflation rate (`inflation_cpi_pct`)?

```python
# Find the row with the maximum 'inflation_cpi_pct'
highest_inflation_year = df.loc[df['inflation_cpi_pct'].idxmax(), ['year', 'inflation_cpi_pct']]
print("Year with the Highest Inflation Rate:")
display(highest_inflation_year)
```

### 3. How has the unemployment rate (`unemployment_pct`) trended over time?

```python
# Select 'year' and 'unemployment_pct' and sort by year
unemployment_trend = df[['year', 'unemployment_pct']].sort_values(by='year').reset_index(drop=True)
print("Unemployment Rate Trend Over Time:")
display(unemployment_trend)
```

4. Which year recorded the largest trade deficit (most negative `trade_balance_usd_bn`)?

```python
# Find the row with the minimum (most negative) 'trade_balance_usd_bn'
largest_trade_deficit = df.loc[df['trade_balance_usd_bn'].idxmin(), ['year', 'trade_balance_usd_bn']]
print("Year with the Largest Trade Deficit:")
display(largest_trade_deficit)
```

### 5. How many years fall into each `gdp_growth_category`?

```python
# Use value_counts() on the 'gdp_growth_category' column
gdp_growth_counts = df['gdp_growth_category'].value_counts().reset_index()
gdp_growth_counts.columns = ['GDP Growth Category', 'Number of Years']
print("Number of Years per GDP Growth Category:")
display(gdp_growth_counts)
```

### 6. What are the top 3 years with the highest foreign exchange reserves (`forex_reserves_usd_bn`)?

```python
# Sort by 'forex_reserves_usd_bn' in descending order and take the top 3
top_forex_reserves = df.nlargest(3, 'forex_reserves_usd_bn')[['year', 'forex_reserves_usd_bn']]
print("Top 3 Years with Highest Foreign Exchange Reserves:")
display(top_forex_reserves)
```

### 7. What is the highest literacy rate (`literacy_rate_pct`) recorded in the dataset?

```python
# Find the maximum value in the 'literacy_rate_pct' column
highest_literacy_rate = df['literacy_rate_pct'].max()
print(f"The highest literacy rate recorded is: {highest_literacy_rate:.2f}%")
```

### 8. What is the average PKR to USD exchange rate (`pkr_per_usd`) across all years?

```python
# Calculate the mean of the 'pkr_per_usd' column
average_exchange_rate = df['pkr_per_usd'].mean()
print(f"The average PKR to USD exchange rate is: {average_exchange_rate:.2f}")
```
