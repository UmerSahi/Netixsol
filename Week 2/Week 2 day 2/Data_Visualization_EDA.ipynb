#Line Graph of Average Disposals per Year

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x='year',y='avg_disposals', errorbar=None)
plt.title('Average Disposals per Year')
plt.xlabel('Year')
plt.ylabel('Average Disposals')
```

#Line Graph of Handballs per Year

```python
plt.figure(figsize=(10,5))
sns.lineplot(data=df, x='year',y='handballs', errorbar=None)
plt.title('Handballs per Year')
plt.xlabel('Year')
plt.ylabel('Handballs')
```

#Line Graph
In this i have have grouped average kicks, average marks, average handballs, average goals and average disposals with year and graph shows these metrics changes over the years

```python
player_stats_yearly = df.groupby('year')[['avg_kicks', 'avg_marks', 'avg_handballs', 'avg_goals', 'avg_disposals']].mean().reset_index()

plt.figure(figsize=(15, 8))
sns.lineplot(data=player_stats_yearly, x='year', y='avg_kicks', label='Average Kicks')
sns.lineplot(data=player_stats_yearly, x='year', y='avg_marks', label='Average Marks')
sns.lineplot(data=player_stats_yearly, x='year', y='avg_handballs', label='Average Handballs')
sns.lineplot(data=player_stats_yearly, x='year', y='avg_goals', label='Average Goals')
sns.lineplot(data=player_stats_yearly, x='year', y='avg_disposals', label='Average Disposals')

plt.title('Average Player Performance Metrics Over Years')
plt.xlabel('Year')
plt.ylabel('Average Value')
plt.legend()
plt.show()
```

#Bar Plot
I have made this chart horizontal and its shows Average kicks per year.

```python
plt.figure(figsize=(10, 12))
sns.barplot(data=df, y='year', x='avg_kicks', errorbar=None, orient='h')
plt.title('Average Kicks per Year')
plt.xlabel('Average Kicks')
plt.ylabel('Year')
plt.show()
```

#Bar Chart
In this I have made a group of team and goals column and the chart shows the number of goals by top 10 teams from Number 1 to Number

```python
goals=df.groupby('team')['goals'].sum().sort_values(ascending=False).head(10).reset_index()
sns.barplot(data=goals, x='team', y='goals',hue='team')
plt.title('Goals per Team')
plt.xlabel('Team')
plt.ylabel('Goals')
plt.xticks(rotation=90)
```

#Scatter plot of average kicks vs Average Marks
In this I have tried to change the colors of both so it can be easily seen.

```python
plt.figure(figsize=(10, 6))
# Plotting kicks and marks with different colors
sns.scatterplot(data=df, x='avg_kicks', y='avg_marks', color='blue', label='Kicks (Blue)', alpha=0.5)
sns.scatterplot(data=df, x='avg_kicks', y='avg_marks', color='red', label='Marks (Red)', alpha=0.3)

plt.title('Scatter Plot of Average Kicks vs Average Marks')
plt.xlabel('Average Kicks')
plt.ylabel('Average Marks')
plt.legend()
plt.show()
```

#HeatMap of some of they key stats of season

```python
plt.figure(figsize=(10, 6))
stat_col=['kicks', 'marks', 'handballs', 'goals', 'disposals','behinds','hit_outs','tackles','one_percenters','total_fantasy_points']
sns.heatmap(df[stat_col].corr(), annot=True, cmap='coolwarm')
plt.title('Correlation of Some Seasonal Stats Heatmap')
plt.show()
```

#In this heatmap is created by using all the numeric columns in our dataset

```python
# Select only numeric columns for correlation analysis
numeric_df = df.select_dtypes(include=['number'])

# Calculate the correlation matrix
correlation_matrix = numeric_df.corr()

# Plotting the heatmap
plt.figure(figsize=(20, 15))
sns.heatmap(correlation_matrix, annot=False, cmap='coolwarm', fmt=".2f", linewidths=.5)
plt.title('Correlation Heatmap of Numeric Performance Columns')
plt.show()
```
