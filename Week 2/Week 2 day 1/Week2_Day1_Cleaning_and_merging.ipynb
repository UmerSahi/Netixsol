```python
# Remove 'ID_' prefix if it exists and convert to integer
stats['player_id'] = stats['player_id'].astype(str).str.replace('ID_', '')
stats['player_id'] = pd.to_numeric(stats['player_id'], errors='coerce').astype(int)
```

```python
stats.isnull().sum()
```

#Kicks Column Missing Values fixed with median

```python
stats['kicks'] = stats['kicks'].fillna(stats['kicks'].median())
```

```python
stats['kicks'].isnull().sum()
```

#Marks Column Missing Values Fixed by Using Median

```python
stats['marks'] = stats['marks'].fillna(stats['marks'].median())
```

```python
stats['marks'].isnull().sum()
```

Again Replacing Hnadballs Column with Median Because Mean is higher and it will then effect the average.

```python
stats['handballs'] = stats['handballs'].fillna(stats['handballs'].median())
```

```python
stats['handballs'].isnull().sum()
```

```python
stats.isnull().sum()
```

#Remaining columns are also replaced by median as mean is not optimal because of outliers in the dataset
so replacing the values of other columns left collectively in on cell

```python
# 1. Filling missing values for common stats using median
columns_to_fix = ['disposals', 'goals', 'behinds', 'tackles', 'hit_outs', 'inside_50s', 'clearances']
for col in columns_to_fix:
    stats[col] = stats[col].fillna(stats[col].median())
```

```python
stats.isnull().sum()
```

We can now check the unique list of teams to ensure they are consistent.

```python
print("Unique Teams:")
print(sorted(stats['team'].unique()))
```

```python
# Convert team names to lowercase and strip whitespace
stats['team'] = stats['team'].str.lower().str.strip()

# Again Verify the unique list of team names
print("Cleaned Unique Teams:")
print(sorted(stats['team'].unique()))
```

```python
#Handle 'disposals' specifically as it is a sum of kicks + handballs
stats['disposals'] = stats['kicks'] + stats['handballs']
```

```python
# Verify results
print("Missing values after cleaning core columns:")
print(stats[columns_to_fix].isnull().sum())
display(stats.head())
```

```python
# Identify columns with 5000 or more missing values
null_counts = stats.isnull().sum()
columns_to_drop_5000 = null_counts[null_counts >= 5000].index.tolist()

print(f"Dropping columns with >= 5000 missing values: {columns_to_drop_5000}")

# Drop the identified columns
stats = stats.drop(columns=columns_to_drop_5000)
```

#filling the remaining columns with median values as well

```python
numeric_columns = stats.select_dtypes(include=["float64", "int64"]).columns

stats[numeric_columns] = stats[numeric_columns].fillna(
    stats[numeric_columns].median()
)
```

#merging the datasets
players dataset id is unique and stats dataset has player_id so we use these 2 to merge and we use inner merge for this.

```python
merged = pd.merge(
    players,
    stats,
    left_on="id",
    right_on="player_id",
    how="inner"
)
```

```python
merged.to_csv("merged_players.csv", index=False)
```
