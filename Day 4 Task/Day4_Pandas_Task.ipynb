```python
import kagglehub

# Download latest version
path = kagglehub.dataset_download("yasserh/titanic-dataset")

print("Path to dataset files:", path)
```

```python
import pandas as pd
import os

# Correcting the dataset file name based on the directory listing
dataset_file_path = f"{path}/Titanic-Dataset.csv"

# Check if the file exists before attempting to read it
if not os.path.exists(dataset_file_path):
    print(f"File not found: {dataset_file_path}. Listing contents of {path}:")
    print(os.listdir(path))
else:
    df = pd.read_csv(dataset_file_path)
```

```python
import numpy as np

# 1. Handle missing values
# Fill missing 'Age' values with the median
df['Age'].fillna(df['Age'].median(), inplace=True)
print("Missing 'Age' values filled with median.")
```

```python
print(df['Age'])
```

```python

# Fill missing 'Embarked' values with the mode (most frequent value)
# .mode()[0] handles cases where there might be multiple modes
df['Embarked'].fillna(df['Embarked'].mode()[0], inplace=True)
print("Missing 'Embarked' values filled with mode.")
```

```python
print(df['Embarked'])
```

```python
# Drop 'Cabin' column due to a large number of missing values
df.drop('Cabin', axis=1, inplace=True)
print("'Cabin' column dropped due to many missing values.")
```

```python
# Verify no more missing values in these columns
print("\nMissing values after handling:")
print(df[['Age', 'Embarked']].isnull().sum())
```

```python
df['Age'].dtype
```

```python
# 3. Remove duplicates
df.drop_duplicates(inplace=True)
print(f"\nDuplicates removed. DataFrame shape after removing duplicates: {df.shape}")
```

```python
# Display the first few rows and info after these operations
print("\nDataFrame head after handling missing values and duplicates:")
print(df.head())
```

```python
print("\nDataFrame info after handling missing values and duplicates:")
df.info()
```

```python
# 4. Create 2 new derived columns
# FamilySize: Sum of SibSp and Parch, plus 1 for the passenger themselves
df['FamilySize'] = df['SibSp'] + df['Parch'] + 1
print("Created 'FamilySize' column.")

# IsAlone: 1 if FamilySize is 1, 0 otherwise
df['IsAlone'] = np.where(df['FamilySize'] == 1, 1, 0)
print("Created 'IsAlone' column.")

# Display new columns
print("\nDataFrame head with new columns:")
print(df[['FamilySize', 'IsAlone']].head())
```

```python
# 5. Reshape the data using .groupby() + .agg() to produce a summary table
# Summarize survival rate, average age, and average fare by Pclass and Sex
summary_table = df.groupby(['Pclass', 'Sex']).agg(
    SurvivalRate=('Survived', 'mean'),
    AverageAge=('Age', 'mean'),
    AverageFare=('Fare', 'mean'),
    TotalPassengers=('PassengerId', 'count')
).reset_index()

print("\nSummary table (Survival rate, avg age, avg fare by Pclass and Sex):")
print(summary_table)
```

```python
# 6. Merge with a second small manually created DataFrame to add extra context
# Create a DataFrame to map Pclass to a more descriptive label
pclass_labels = pd.DataFrame({
    'Pclass': [1, 2, 3],
    'ClassDescription': ['First Class', 'Second Class', 'Third Class']
})

# Merge the summary table with the Pclass labels
summary_with_context = pd.merge(summary_table, pclass_labels, on='Pclass', how='left')

print("\nSummary table after merging with Class Description:")
print(summary_with_context)

# Display overall info after all operations
print("\nFinal DataFrame info after all operations:")
df.info()
```
